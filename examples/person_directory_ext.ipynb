## Person Directory Extension


```python
from base import _init_env, get_config

from azure_ai_vision_person import PersonDirectoryExtension

_init_env()
```

```python
pde = PersonDirectoryExtension(*get_config())
```

```python
rsp = await pde.get_actual_persons()

rsp
```

```python
import json


def age(user_data: str) -> int:
    return json.loads(user_data).get("age")


await pde.get_actual_persons(where=lambda p: age(str(p.user_data)) == 22)
```
