## Dynamic Person Group Client


```python
from azure_ai_vision_person import DynamicPersonGroup

dpg = DynamicPersonGroup(*get_config())
```

### List available groups


```python
await dpg.get_groups()
```

### Create group


#### Get persons to add


```python
from azure_ai_vision_person import PersonDirectoryExtension

pde = PersonDirectoryExtension(*get_config())
```

```python
persons = await pde.get_persons()
```

```python
await dpg.create_group(
    "test_group",
    name="class1",
    user_data={"class_no": 1},
    person_ids=[p.person_id for p in persons.persons],
)
```

#### Verify creation of dynamic person group

```python
await dpg.get_groups()
```
